```python
import pandas as pd

loans = pd.read_csv('/content/drive/MyDrive/kiva_loans.csv')

loans.head()
```

```python
loans.info()
```

```python
loans['date'] = pd.to_datetime(loans['date'])

print("Earliest date:", loans['date'].min())
print("Latest date:", loans['date'].max())
```

# Q1: Which five sectors received the highest total funding in the first two years of the dataset (2014–2015)?

```python
# Keep loans from the first two years
first_two_years = loans[loans['date'].dt.year.isin([2014, 2015])]


first_two_years.head()
```

```python
# Add up funding for each sector
sector_funding = first_two_years.groupby('sector')['funded_amount'].sum()

# Sort from highest to lowest and show the top five
top_five_sectors = sector_funding.sort_values(ascending=False).head(5)

top_five_sectors
```

The analysis shows that agriculture sector received the most funding from 2014 to 2015. The food sector is second with a difference of about 10.5 million

# Q2: Which five countries have the highest number of loan requests in the dataset?

```python
# Count loan records for each country

country_requests = loans.groupby('country').size()

# Show the five highest counts
top_five_countries = country_requests.sort_values(ascending=False).head(5)

top_five_countries

```

In this dataset, from 2014 to 2017, Philippines, Kenya, El Salvador, Cambodia and Pakistan are the top five countries with the highest number of loan requests. Kenya ranks second, with slightly less than half as many loan requests as the Philippines.

# Q3: Which five individual loans have the largest gaps between requested and funded amounts?

```python
# Calculate the funding gap for each loan

loans['funding_gap'] = loans.apply(lambda row: row['loan_amount'] - row['funded_amount'], axis=1)

loans[['id', 'loan_amount', 'funded_amount', 'funding_gap']].head()
```

```python
# Sort loans by funding gap with the largest first

largest_gaps = loans.sort_values('funding_gap', ascending=False)

# Show the five largest gaps
largest_gaps[['id', 'country', 'sector', 'loan_amount', 'funded_amount', 'funding_gap']].head(5)
```

In the dataset, the five largest funding gaps range from 40,025 to 50,000. A Wholesale loan in Haiti and a Transportation loan in Mexico tie for the largest gap at 50,000 each. The remaining three loans are in Food and Agriculture.

```python
# Find loan descriptions mentioning fertilizer
fertilizer_loans = loans[
loans['use'].str.contains('fertilizer', case=False, na=False)]


# Keep requests of at least 5,000
large_fertilizer_loans = fertilizer_loans[
fertilizer_loans['loan_amount'] >= 5000]


# Count loans for each country
fertilizer_countries = large_fertilizer_loans.groupby('country').size()

# Five countries with the highest counts
fertilizer_countries.sort_values(ascending=False).head(5)
```

In the dataset, all loan requests of at least 5,000 dollars which had descriptions mention fertilizer, Peru has the most loan requests with 62, followed by Kenya with 39. Myanmar (Burma), Guatemala and Honduras complete the top five, with 33, 21 and 11 loan requests respectively.

# Q5: How does the average scheduled repayment duration vary between countries?

```python
# Calculate the average scheduled repayment duration for each country
country_duration = loans.groupby('country')['term_in_months'].mean()

# Show all countries, from longest to shortest average duration
country_duration.sort_values(ascending=False)
```

In the dataset, the average scheduled repayment duration is different across the 87 countries in the dataset. Meanwhile, South Africa has the longest average duration at approximately 53 months, followed by Belize at 51 months. Congo has the shortest average duration at approximately 6.4 months.

# Q6: Which repayment schedule is most common among loan requests in the dataset?

```python
# Count loan requests for each repayment schedule
schedule_counts = loans.groupby('repayment_interval').size()

# Show the schedules from most to least common
schedule_counts.sort_values(ascending=False)
```

In the dataset, monthly repayment is the most common repayment schedule, with 342,717 loan requests, followed by irregular with 257,158 loan requests. Weekly repayment which is the least common, with 602 loan requests.
